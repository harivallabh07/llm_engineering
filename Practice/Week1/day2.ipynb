```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()  # Load environment variables from .env file
gemini_api_key=os.getenv("GEMINI_API_KEY")  # Get the Google API key from environment variables
gemini_base_url=os.getenv("GEMINI_BASE_URL")  # Get the Google Base URL from environment variables
```

```python
message = "What is the capital of France?"
messages=[{'role':'user','content':message}]
messages
```

```python
gemini = OpenAI(base_url=gemini_base_url, api_key=gemini_api_key)
response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=messages)
response.choices[0].message.content
```
