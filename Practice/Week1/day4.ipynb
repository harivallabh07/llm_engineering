```python
import tiktoken 
encoding=tiktoken.encoding_for_model('gpt-4-mini')
tokens=encoding.encode('Hi, My name is Hari and I am learning LLM engineering.')
for token in tokens:
    print(f'{token} --> {encoding.decode([token])} ')
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
gemini_base_url=os.getenv('GEMINI_BASE_URL')
gemini_api_key=os.getenv('GEMINI_API_KEY')
```

```python
gemini=OpenAI(base_url=gemini_base_url, api_key=gemini_api_key)
```

```python
messages=[{"role": "user", "content": "Hi, My name is Hari"}]
```

```python
response=gemini.chat.completions.create(model='gemini-2.5-flash',messages=messages)
print(response.choices[0].message.content)
```

```python
messages=[{"role": "system", "content": "U are a helpful assistant."},
          {"role": "user", "content": "Genearte an image of a cat."}]
response=gemini.chat.completions.create(model='gemini-2.5-flash', messages=messages)
print(response.choices[0].message.content)
```
